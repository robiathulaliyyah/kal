# Matriks Transformasi dari Grafik

## Konsep Translasi
Transformasi translasi memindahkan titik:
$$
(x, y) \rightarrow (x+a, y+b)
$$

Bentuk matriks homogen:
$$
\begin{bmatrix}
x' \\ y' \\ 1
\end{bmatrix}
$$

=

$$
\begin{bmatrix}
1 & 0 & a \\
0 & 1 & b \\
0 & 0 & 1
\end{bmatrix}
$$

$$
\begin{bmatrix}
x \\ y \\ 1
\end{bmatrix}
$$

---

## 1. Transformasi A → B

Diketahui:
$$
A(2,3), \ B(2,1)
$$

### Langkah:
$$
a = 2 - 2 = 0 
$$
$$
b = 1 - 3 = -2
$$

### Vektor Translasi:
$$
(0, -2)
$$

### Matriks Transformasi:
$$
T_{AB} =
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & -2 \\
0 & 0 & 1
\end{bmatrix}
$$

### Verifikasi:
$$
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & -2 \\
0 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
2 \\ 3 \\ 1
\end{bmatrix}
$$
=

$$
\begin{bmatrix}
2 \\ 1 \\ 1
\end{bmatrix}
= B
$$
