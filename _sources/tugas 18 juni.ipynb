## A. Soal Determinan Matriks 3×3

### Soal 1
Hitung determinan dari matriks berikut:

$$
A=
\begin{bmatrix}
2 & 1 & 3 \\
0 & 4 & 5 \\
1 & 2 & 1
\end{bmatrix}
$$

### Penyelesaian

Gunakan rumus determinan matriks 3×3:

$$
\det(A)
=
a\det(A_{11})
-b\det(A_{12})
+c\det(A_{13})
$$

dengan

$$
A=
\begin{bmatrix}
a & b & c \\
d & e & f \\
g & h & i
\end{bmatrix}
$$

Pada soal:

$$
a=2,\quad b=1,\quad c=3
$$

$$
d=0,\quad e=4,\quad f=5
$$

$$
g=1,\quad h=2,\quad i=1
$$

Substitusikan ke rumus:

$$
\det(A)
=
2
\begin{vmatrix}
4 & 5 \\
2 & 1
\end{vmatrix}
-
1
\begin{vmatrix}
0 & 5 \\
1 & 1
\end{vmatrix}
+
3
\begin{vmatrix}
0 & 4 \\
1 & 2
\end{vmatrix}
$$

Hitung masing-masing determinan 2×2:

$$
\begin{vmatrix}
4 & 5 \\
2 & 1
\end{vmatrix}
=
(4)(1)-(5)(2)
=
4-10
=
-6
$$

$$
\begin{vmatrix}
0 & 5 \\
1 & 1
\end{vmatrix}
=
(0)(1)-(5)(1)
=
-5
$$

$$
\begin{vmatrix}
0 & 4 \\
1 & 2
\end{vmatrix}
=
(0)(2)-(4)(1)
=
-4
$$

Sehingga:

$$
\det(A)
=
2(-6)-1(-5)+3(-4)
$$

$$
=
-12+5-12
$$

$$
=
-19
$$

### Jawaban

$$
\boxed{\det(A)=-19}
$$

## Soal 3

Jika

$$
C=
\begin{bmatrix}
1 & 2 & 3 \\
2 & 4 & 6 \\
1 & 1 & 1
\end{bmatrix}
$$

tentukan determinan matriks \(C\) dan jelaskan apakah matriks tersebut singular atau tidak.

### Penyelesaian

Gunakan rumus determinan matriks 3×3:

$$
\det(C)
=
a\det(C_{11})
-b\det(C_{12})
+c\det(C_{13})
$$

dengan

$$
C=
\begin{bmatrix}
a & b & c \\
d & e & f \\
g & h & i
\end{bmatrix}
$$

Pada soal:

$$
a=1,\quad b=2,\quad c=3
$$

$$
d=2,\quad e=4,\quad f=6
$$

$$
g=1,\quad h=1,\quad i=1
$$

Substitusikan ke rumus:

$$
\det(C)
=
1
\begin{vmatrix}
4 & 6 \\
1 & 1
\end{vmatrix}
-
2
\begin{vmatrix}
2 & 6 \\
1 & 1
\end{vmatrix}
+
3
\begin{vmatrix}
2 & 4 \\
1 & 1
\end{vmatrix}
$$

Hitung masing-masing determinan 2×2:

$$
\begin{vmatrix}
4 & 6 \\
1 & 1
\end{vmatrix}
=
(4)(1)-(6)(1)
=
-2
$$

$$
\begin{vmatrix}
2 & 6 \\
1 & 1
\end{vmatrix}
=
(2)(1)-(6)(1)
=
-4
$$

$$
\begin{vmatrix}
2 & 4 \\
1 & 1
\end{vmatrix}
=
(2)(1)-(4)(1)
=
-2
$$

Sehingga:

$$
\det(C)
=
1(-2)-2(-4)+3(-2)
$$

$$
=
-2+8-6
$$

$$
=
0
$$

### Kesimpulan

Karena

$$
\det(C)=0
$$

maka matriks \(C\) adalah **matriks singular**.

Matriks singular adalah matriks yang determinannya sama dengan nol sehingga **tidak memiliki invers**.

Selain itu, dapat dilihat bahwa:

$$
\text{Baris ke-2} = 2 \times \text{Baris ke-1}
$$

yaitu

$$
[2\;\;4\;\;6]
=
2[1\;\;2\;\;3]
$$

Karena ada baris yang merupakan kelipatan baris lain, baris-baris matriks tidak independen sehingga determinannya bernilai nol.

### Jawaban

$$
\boxed{\det(C)=0}
$$

Karena determinannya nol, maka:

$$
\boxed{\text{Matriks } C \text{ adalah matriks singular}}
$$

## B. Soal Dekomposisi Matriks (LU Decomposition)

### Soal 4

Lakukan dekomposisi LU terhadap matriks:

$$
A=
\begin{bmatrix}
2 & 4 & 2 \\
1 & 5 & 2 \\
1 & 2 & 4
\end{bmatrix}
$$

dengan bentuk:

$$
A = LU
$$

di mana \(L\) adalah matriks segitiga bawah dan \(U\) adalah matriks segitiga atas.

---

### Penyelesaian

Misalkan

$$
L=
\begin{bmatrix}
1 & 0 & 0 \\
l_{21} & 1 & 0 \\
l_{31} & l_{32} & 1
\end{bmatrix}
$$

dan

$$
U=
\begin{bmatrix}
u_{11} & u_{12} & u_{13} \\
0 & u_{22} & u_{23} \\
0 & 0 & u_{33}
\end{bmatrix}
$$

Karena

$$
A=LU
$$

maka elemen-elemen \(L\) dan \(U\) dicari menggunakan eliminasi Gauss.

---

### Langkah 1: Menentukan baris pertama U

Dari baris pertama matriks \(A\):

$$
u_{11}=2,\qquad
u_{12}=4,\qquad
u_{13}=2
$$

sehingga

$$
U=
\begin{bmatrix}
2 & 4 & 2 \\
0 & * & * \\
0 & 0 & *
\end{bmatrix}
$$

---

### Langkah 2: Menentukan elemen kolom pertama L

$$
l_{21}
=
\frac{a_{21}}{u_{11}}
=
\frac{1}{2}
$$

$$
l_{31}
=
\frac{a_{31}}{u_{11}}
=
\frac{1}{2}
$$

Maka

$$
L=
\begin{bmatrix}
1 & 0 & 0 \\
\frac12 & 1 & 0 \\
\frac12 & * & 1
\end{bmatrix}
$$

---

### Langkah 3: Eliminasi baris kedua

Baris 2 baru:

$$
R_2-\frac12R_1
$$

$$
=
\begin{bmatrix}
1 & 5 & 2
\end{bmatrix}
-
\frac12
\begin{bmatrix}
2 & 4 & 2
\end{bmatrix}
$$

$$
=
\begin{bmatrix}
0 & 3 & 1
\end{bmatrix}
$$

Sehingga

$$
u_{22}=3,
\qquad
u_{23}=1
$$

---

### Langkah 4: Eliminasi baris ketiga

Baris 3 baru:

$$
R_3-\frac12R_1
$$

$$
=
\begin{bmatrix}
1 & 2 & 4
\end{bmatrix}
-
\frac12
\begin{bmatrix}
2 & 4 & 2
\end{bmatrix}
$$

$$
=
\begin{bmatrix}
0 & 0 & 3
\end{bmatrix}
$$

Untuk menghilangkan elemen di bawah pivot kedua:

$$
l_{32}
=
\frac{0}{3}
=
0
$$

dan

$$
u_{33}=3
$$

---

### Matriks L

$$
L=
\begin{bmatrix}
1 & 0 & 0 \\
\frac12 & 1 & 0 \\
\frac12 & 0 & 1
\end{bmatrix}
$$

### Matriks U

$$
U=
\begin{bmatrix}
2 & 4 & 2 \\
0 & 3 & 1 \\
0 & 0 & 3
\end{bmatrix}
$$

---

### Verifikasi

$$
LU=
\begin{bmatrix}
1 & 0 & 0 \\
\frac12 & 1 & 0 \\
\frac12 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
2 & 4 & 2 \\
0 & 3 & 1 \\
0 & 0 & 3
\end{bmatrix}
=
\begin{bmatrix}
2 & 4 & 2 \\
1 & 5 & 2 \\
1 & 2 & 4
\end{bmatrix}
=A
$$

---

### Jawaban

$$
L=
\begin{bmatrix}
1 & 0 & 0 \\
\frac12 & 1 & 0 \\
\frac12 & 0 & 1
\end{bmatrix}
$$

$$
U=
\begin{bmatrix}
2 & 4 & 2 \\
0 & 3 & 1 \\
0 & 0 & 3
\end{bmatrix}
$$

Sehingga

$$
\boxed{
A=
\begin{bmatrix}
1 & 0 & 0 \\
\frac12 & 1 & 0 \\
\frac12 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
2 & 4 & 2 \\
0 & 3 & 1 \\
0 & 0 & 3
\end{bmatrix}
}
$$

## Soal 5

Tentukan matriks \(L\) dan \(U\) dari:

$$
B=
\begin{bmatrix}
1 & 2 & 1 \\
2 & 5 & 3 \\
4 & 10 & 8
\end{bmatrix}
$$

menggunakan metode eliminasi Gauss.

---

### Penyelesaian

Kita akan melakukan eliminasi Gauss pada matriks \(B\) hingga menjadi matriks segitiga atas \(U\).

Misalkan:

$$
B=
\begin{bmatrix}
1 & 2 & 1 \\
2 & 5 & 3 \\
4 & 10 & 8
\end{bmatrix}
$$

---

### Langkah 1: Eliminasi elemen di bawah pivot pertama

Pivot pertama:

$$
a_{11}=1
$$

Faktor pengali untuk baris kedua:

$$
m_{21}=\frac{2}{1}=2
$$

Lakukan:

$$
R_2 \leftarrow R_2-2R_1
$$

$$
=
\begin{bmatrix}
2 & 5 & 3
\end{bmatrix}
-
2
\begin{bmatrix}
1 & 2 & 1
\end{bmatrix}
=
\begin{bmatrix}
0 & 1 & 1
\end{bmatrix}
$$

Faktor pengali untuk baris ketiga:

$$
m_{31}=\frac{4}{1}=4
$$

Lakukan:

$$
R_3 \leftarrow R_3-4R_1
$$

$$
=
\begin{bmatrix}
4 & 10 & 8
\end{bmatrix}
-
4
\begin{bmatrix}
1 & 2 & 1
\end{bmatrix}
=
\begin{bmatrix}
0 & 2 & 4
\end{bmatrix}
$$

Matriks menjadi:

$$
\begin{bmatrix}
1 & 2 & 1 \\
0 & 1 & 1 \\
0 & 2 & 4
\end{bmatrix}
$$

---

### Langkah 2: Eliminasi elemen di bawah pivot kedua

Pivot kedua:

$$
a_{22}=1
$$

Faktor pengali:

$$
m_{32}=\frac{2}{1}=2
$$

Lakukan:

$$
R_3 \leftarrow R_3-2R_2
$$

$$
=
\begin{bmatrix}
0 & 2 & 4
\end{bmatrix}
-
2
\begin{bmatrix}
0 & 1 & 1
\end{bmatrix}
=
\begin{bmatrix}
0 & 0 & 2
\end{bmatrix}
$$

Sehingga diperoleh matriks segitiga atas:

$$
U=
\begin{bmatrix}
1 & 2 & 1 \\
0 & 1 & 1 \\
0 & 0 & 2
\end{bmatrix}
$$

---

### Menentukan Matriks L

Matriks \(L\) dibentuk dari faktor-faktor eliminasi:

$$
m_{21}=2,
\qquad
m_{31}=4,
\qquad
m_{32}=2
$$

dan diagonal utama bernilai 1.

$$
L=
\begin{bmatrix}
1 & 0 & 0 \\
2 & 1 & 0 \\
4 & 2 & 1
\end{bmatrix}
$$

---

### Verifikasi

$$
LU=
\begin{bmatrix}
1 & 0 & 0 \\
2 & 1 & 0 \\
4 & 2 & 1
\end{bmatrix}
\begin{bmatrix}
1 & 2 & 1 \\
0 & 1 & 1 \\
0 & 0 & 2
\end{bmatrix}
=
\begin{bmatrix}
1 & 2 & 1 \\
2 & 5 & 3 \\
4 & 10 & 8
\end{bmatrix}
=B
$$

---

### Jawaban

$$
L=
\begin{bmatrix}
1 & 0 & 0 \\
2 & 1 & 0 \\
4 & 2 & 1
\end{bmatrix}
$$

$$
U=
\begin{bmatrix}
1 & 2 & 1 \\
0 & 1 & 1 \\
0 & 0 & 2
\end{bmatrix}
$$

Sehingga:

$$
\boxed{
B=
\begin{bmatrix}
1 & 0 & 0 \\
2 & 1 & 0 \\
4 & 2 & 1
\end{bmatrix}
\begin{bmatrix}
1 & 2 & 1 \\
0 & 1 & 1 \\
0 & 0 & 2
\end{bmatrix}
}
$$

## Soal 6

C=

$$

$$
=
\begin{bmatrix}

\end{bmatrix}

\begin{bmatrix}

$$
\begin{bmatrix}

$$
\begin{bmatrix}

## C. Soal Invers Matriks 3×3

### Soal 7

Tentukan invers dari matriks:

$$
A=
\begin{bmatrix}
1 & 2 & 1 \\
0 & 1 & 1 \\
2 & 3 & 4
\end{bmatrix}
$$

menggunakan metode Gauss-Jordan.

---

### Penyelesaian

Bentuk matriks augmented \([A|I]\):

$$
\left[
\begin{array}{ccc|ccc}
1 & 2 & 1 & 1 & 0 & 0 \\
0 & 1 & 1 & 0 & 1 & 0 \\
2 & 3 & 4 & 0 & 0 & 1
\end{array}
\right]
$$

Tujuan metode Gauss-Jordan adalah mengubah bagian kiri menjadi matriks identitas.

---

### Langkah 1

Hilangkan elemen di bawah pivot pertama.

$$
R_3 \leftarrow R_3 - 2R_1
$$

$$
\left[
\begin{array}{ccc|ccc}
1 & 2 & 1 & 1 & 0 & 0 \\
0 & 1 & 1 & 0 & 1 & 0 \\
0 & -1 & 2 & -2 & 0 & 1
\end{array}
\right]
$$

---

### Langkah 2

Hilangkan elemen di bawah pivot kedua.

$$
R_3 \leftarrow R_3 + R_2
$$

$$
\left[
\begin{array}{ccc|ccc}
1 & 2 & 1 & 1 & 0 & 0 \\
0 & 1 & 1 & 0 & 1 & 0 \\
0 & 0 & 3 & -2 & 1 & 1
\end{array}
\right]
$$

---

### Langkah 3

Buat pivot ketiga bernilai 1.

$$
R_3 \leftarrow \frac{1}{3}R_3
$$

$$
\left[
\begin{array}{ccc|ccc}
1 & 2 & 1 & 1 & 0 & 0 \\
0 & 1 & 1 & 0 & 1 & 0 \\
0 & 0 & 1 &
-\frac{2}{3} &
\frac{1}{3} &
\frac{1}{3}
\end{array}
\right]
$$

---

### Langkah 4

Hilangkan elemen di atas pivot ketiga.

$$
R_2 \leftarrow R_2 - R_3
$$

$$
R_1 \leftarrow R_1 - R_3
$$

Hasil:

$$
\left[
\begin{array}{ccc|ccc}
1 & 2 & 0 &
\frac{5}{3} &
-\frac{1}{3} &
-\frac{1}{3}
\\
0 & 1 & 0 &
\frac{2}{3} &
\frac{2}{3} &
-\frac{1}{3}
\\
0 & 0 & 1 &
-\frac{2}{3} &
\frac{1}{3} &
\frac{1}{3}
\end{array}
\right]
$$

---

### Langkah 5

Hilangkan elemen di atas pivot kedua.

$$
R_1 \leftarrow R_1 - 2R_2
$$

$$
\left[
\begin{array}{ccc|ccc}
1 & 0 & 0 &
\frac{1}{3} &
-\frac{5}{3} &
\frac{1}{3}
\\
0 & 1 & 0 &
\frac{2}{3} &
\frac{2}{3} &
-\frac{1}{3}
\\
0 & 0 & 1 &
-\frac{2}{3} &
\frac{1}{3} &
\frac{1}{3}
\end{array}
\right]
$$

Karena bagian kiri telah menjadi matriks identitas, maka bagian kanan adalah invers matriks \(A\).

---

### Jawaban

$$
A^{-1}
=
\begin{bmatrix}
\frac{1}{3} & -\frac{5}{3} & \frac{1}{3}
\\[4pt]
\frac{2}{3} & \frac{2}{3} & -\frac{1}{3}
\\[4pt]
-\frac{2}{3} & \frac{1}{3} & \frac{1}{3}
\end{bmatrix}
$$

atau dapat ditulis sebagai:

$$
A^{-1}
=
\frac{1}{3}
\begin{bmatrix}
1 & -5 & 1 \\
2 & 2 & -1 \\
-2 & 1 & 1
\end{bmatrix}
$$

$$
\boxed{
A^{-1}
=
\frac{1}{3}
\begin{bmatrix}
1 & -5 & 1 \\
2 & 2 & -1 \\
-2 & 1 & 1
\end{bmatrix}
}
$$
